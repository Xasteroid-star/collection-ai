##广播机制与向量化距离计算

```python
import numpy as np

points_A = np.random.randint(0, 101, (5, 2))
points_B = np.random.randint(0, 101, (8, 2))

# 2. 距离矩阵计算 
diff = points_A[:, None, :] - points_B[None, :, :]
distance_matrix = np.sqrt(np.sum(diff**2, axis=2))

# 3. 轴向数据聚合
min_distances = np.min(distance_matrix, axis=1)

# 4. 复合条件查询
mask = np.any(distance_matrix < 20, axis=0)
indices = np.where(mask)[0]

# 打印结果供参考
print( points_A)
print( points_B)
print( distance_matrix)
print( min_distances)
print(indices)
```
