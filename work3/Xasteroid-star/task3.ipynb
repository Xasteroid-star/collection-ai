##作业 3：矩阵操作与数据分析

```python
import numpy as np

# 数组创建
data_matrix = np.arange(100).reshape(10, 10).astype(float)

# 子矩阵提取
sub_matrix = data_matrix[3:7, 3:7]

# 条件筛选与赋值
data_matrix[data_matrix > 75] = 0

#  向量化运算
data_matrix *= 0.8

# 数据聚合与定位
max_val = data_matrix.max()
max_index = np.unravel_index(data_matrix.argmax(), data_matrix.shape)

print(f"最终矩阵中的最大值: {max_val}")
print(f"最大值的行索引和列索引: {max_index}")
```
