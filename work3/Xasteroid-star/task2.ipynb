作业 2：Logistic 函数（Matplotlib）

```python
import numpy as np

import matplotlib.pyplot as plt

def logistic(x, L, k, x0):
    return L / (1 + np.exp(-k * (x - x0)))

x = np.linspace(-10, 10, 400)

fig, axes = plt.subplots(2, 2, figsize=(12, 10))
fig.suptitle('Logistic Function Parameters Analysis', fontsize=16)

# Standard Logistic Function (L=1, k=1, x0=0)
axes[0, 0].plot(x, logistic(x, 1, 1, 0), 'r', label='L=1, k=1, x0=0')
axes[0, 0].set_title('Standard Logistic (Sigmoid)')
axes[0, 0].grid(True)
axes[0, 0].legend()

# Effect of L (Maximum value)
for L_val in [0.5, 1.0, 1.5]:
    axes[0, 1].plot(x, logistic(x, L_val, 1, 0), label=f'L={L_val}')
axes[0, 1].set_title('Effect of L: Changes Saturation Value')
axes[0, 1].grid(True)
axes[0, 1].legend()

# Effect of k (Steepness)
for k_val in [0.5, 1.0, 2.0]:
    axes[1, 0].plot(x, logistic(x, 1, k_val, 0), label=f'k={k_val}')
axes[1, 0].set_title('Effect of k: Changes Steepness')
axes[1, 0].grid(True)
axes[1, 0].legend()

# Effect of x0 (Midpoint shift)
for x0_val in [-3, 0, 3]:
    axes[1, 1].plot(x, logistic(x, 1, 1, x0_val), label=f'x0={x0_val}')
axes[1, 1].set_title('Effect of x0: Shifts Curve Horizontally')
axes[1, 1].grid(True)
axes[1, 1].legend()

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

L决定函数的最大值

K决定函数变化速率 

x0决定函数的左右位置
